```python
# Setup for ChatGPT API calls
import openai
import os
import json
from dotenv import load_dotenv

# Load environment variables
load_dotenv()
openai.api_key = os.getenv("OPENAI_API_KEY")

def query_chatgpt(prompt, requirement_num):
    """
    Query ChatGPT with a specific prompt for a requirement
    """
    try:
        response = openai.ChatCompletion.create(
            model="gpt-3.5-turbo",
            messages=[
                {"role": "user", "content": prompt}
            ],
            temperature=0
        )
        return response.choices[0].message.content
    except Exception as e:
        return f"API call failed: {e}"

print("ChatGPT API setup complete")
print("Ready to query for Python Milvus implementations")
```
