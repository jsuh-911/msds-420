````python
# Analysis of ASIN relationship between metadata and reviews DataFrames
print("=== ASIN Analysis: Metadata vs Reviews ===\n")

# Get unique ASINs from each DataFrame
metadata_asins = set(appliances_metadata_df['asin'].dropna())
reviews_asins = set(appliances_reviews_df['asin'].dropna())

print(f"Unique ASINs in metadata: {len(metadata_asins)}")
print(f"Unique ASINs in reviews: {len(reviews_asins)}")

# Check overlap
common_asins = metadata_asins.intersection(reviews_asins)
metadata_only = metadata_asins - reviews_asins
reviews_only = reviews_asins - metadata_asins

print(f"\nASIN overlap analysis:")
print(f"ASINs in both datasets: {len(common_asins)}")
print(f"ASINs only in metadata: {len(metadata_only)}")
print(f"ASINs only in reviews: {len(reviews_only)}")

# 1:1 relationship check
is_perfect_match = len(metadata_asins) == len(reviews_asins) == len(common_asins)
print(f"\nPerfect 1:1 ASIN match: {is_perfect_match}")

# Coverage percentages
metadata_coverage = len(common_asins) / len(metadata_asins) * 100 if metadata_asins else 0
reviews_coverage = len(common_asins) / len(reviews_asins) * 100 if reviews_asins else 0

print(f"\nCoverage analysis:")
print(f"Metadata ASINs with reviews: {metadata_coverage:.1f}%")
print(f"Review ASINs with metadata: {reviews_coverage:.1f}%")

# Show examples of non-matching ASINs
if metadata_only:
    print(f"\nExample ASINs only in metadata (first 5): {list(metadata_only)[:5]}")
if reviews_only:
    print(f"Example ASINs only in reviews (first 5): {list(reviews_only)[:5]}")

print(f"\n=== SUMMARY ===")
print(f"Total unique ASINs across both datasets: {len(metadata_asins.union(reviews_asins))}")
print(f"Relationship type: {'Perfect 1:1 match' if is_perfect_match else 'Not 1:1 - datasets have different ASIN coverage'}")
````

````python
from openai import OpenAI
import os
import json
import requests
from dotenv import load_dotenv  # Add this import
from tenacity import retry, wait_random_exponential, stop_after_attempt
from termcolor import colored

# Load environment variables
load_dotenv()  

# Updated model 
GPT_MODEL = "gpt-3.5-turbo"

# SECURE: Use environment variable
client = OpenAI(
    api_key=os.getenv("OPENAI_API_KEY")
)

@retry(wait=wait_random_exponential(min=1, max=60), stop=stop_after_attempt(6))
def get_completion(prompt, model=GPT_MODEL, temperature=0):
    """Get completion with retry logic and error handling"""
    try:
        messages = [{"role": "user", "content": prompt}]
        response = client.chat.completions.create(
            model=model,
            messages=messages,
            temperature=temperature,
        )
        return response.choices[0].message.content
    except Exception as e:
        print(f"Error: {e}")
        return None

# Test the function
def test_connection():
    response = get_completion("Say this is a test")
    print(response)

# Run the test
test_connection()

````
