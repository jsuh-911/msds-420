````python
# Load the JSON data from a file
file_path = 'chicago_yelp_reviews.json'

# Open the file and read line by line
with open(file_path, 'r') as file:
    for line in file:
        # Parse each line as JSON and append to the list
        list__business_reviews_documents.append(json.loads(line))

# Convert the list of dictionaries to a DataFrame
df= pd.DataFrame(list__business_reviews_documents)


source_df = pd.json_normalize(df['source'])


df_businesses = df.drop(columns=['id','source']).join(source_df)
````

### Requirement 1:
- Create a Graph-Relational Data Model for Yelp Businesses/Reviews using **geldata/edgedb**.
- Insert the YELP REVIEW objects into the geldata/edgedb database.

Consider the following code snippet for your schema:

```
module default {
    type Country {
        property name -> str;
        multi link has_states := .<in_country[is State];
        constraint exclusive on (.name);
    }

    type State {
        property name -> str;
        link in_country -> Country;
        multi link has_cities := .<in_state[is City];
        constraint exclusive on ((.name, .in_country));
    }

    type City {
        property name -> str;
        link in_state -> State;
        multi link has_communityAreas := .<in_city[is CommunityArea];
        multi link has_zipCodes := .<in_city[is ZipCode];
        constraint exclusive on ((.name, .in_state));
    }

    type CommunityArea {
        property name -> str;
        link in_city -> City;
        link has_zipcodes := .<in_communityArea[is ZipCode];
        multi link has_businesses := .<in_communityArea[is Business];
        constraint exclusive on ((.name, .in_city));
    }

    type ZipCode {
        property digits -> str;
        link in_city -> City;
        multi link in_communityArea -> CommunityArea;
        multi link has_streetAddresses := .<in_zipcode[is StreetAddress];
        constraint exclusive on (.digits);
    }

    type StreetAddress {
        property address1 -> str;
        property address2 -> str;
        property address3 -> str;
        link in_zipcode -> ZipCode;
        property coordinates -> tuple<latitude: float32, longitude: float32>;
        link has_business := .<has_address[is Business];
        constraint exclusive on (.coordinates);
    }

    type Business {
        property ID -> str;
        property name -> str;
        property alias -> str;
        property rating -> float32;
        property review_count -> int32;
        link in_communityArea -> CommunityArea;
        property coordinates -> tuple<latitude: float32, longitude: float32>;
        link has_address -> StreetAddress;
        multi link in_category -> Category;
        multi link has_reviews := .<reviews_business[is Review];
        constraint exclusive on (.ID);
    }

    type Category {
        property name -> str;
        link has_business := .<in_category[is Business];
        constraint exclusive on (.name);
    }

    type Review {
	    property review_id -> str;
        property text -> str;
        property rating -> float32;
        property time_created -> str;
        link reviews_business -> Business;
        link written_by -> User;
    }

    type User {
        property user_id -> str;
        multi link reviews := .<written_by[is Review];
        constraint exclusive on (.user_id);
    }
}
                                              
```

````python
import edgedb
import time


def find_pizza_restaurants():
    """
    Find pizza businesses in Chicago zip codes.
    """
    # Connect to database
    client = edgedb.create_client()
    
    # Chicago zip codes to search
    zip_codes = ['60601', '60602', '60603', '60604', '60605', '60606']
    category = 'Pizza'
    
    print(f"Looking for {category} businesses in Chicago zip codes...")
    print()
    
    # Start timing
    start_time = time.time()
    
    # Query to find pizza businesses
    query = """
        SELECT Business {
            name,
            rating,
            full_address := .has_address.address1 ++ ', ' ++
                           .has_address.in_zipcode.in_city.name ++ ', ' ++
                           .has_address.in_zipcode.in_city.in_state.name ++ ' ' ++
                           .has_address.in_zipcode.digits
        }
        FILTER
            EXISTS (SELECT .in_category FILTER .name = <str>$category)
            AND
            .has_address.in_zipcode.digits IN array_unpack(<array<str>>$zip_codes);
    """
    
    # Run the query
    results = client.query(query, category=category, zip_codes=zip_codes)
    
    # Stop timing
    end_time = time.time()
    total_time = end_time - start_time
    
    # Show results
    print(f"Found {len(results)} pizza businesses:")
    print()
    
    for business in results:
        print(f"- {business.name}")
        print(f"  Rating: {business.rating}")
        print(f"  Address: {business.full_address}")
        print()
    
    print(f"Query took {total_time:.3f} seconds")
    
    # Close connection
    client.close()


if __name__ == "__main__":
    find_pizza_restaurants()
````

````python
import edgedb
import time


def find_italian_restaurants_with_reviews():
    """
    Find top 3 Italian restaurants in zip code 60601 and get 3 reviews for each.
    """
    # Connect to database
    client = edgedb.create_client()
    
    # What we're looking for
    zip_code = '60601'
    category = 'Italian'
    
    print(f"Finding top 3 {category} restaurants in zip code {zip_code}")
    print("Then getting 3 reviews for each restaurant...")
    print()
    
    # Start timing
    start_time = time.time()
    
    # First, find the top 3 Italian restaurants
    business_query = """
        SELECT Business {
            ID,
            name,
            review_count,
            rating,
            full_address := .has_address.address1 ++ ', ' ++
                           .has_address.in_zipcode.in_city.name ++ ', ' ++
                           .has_address.in_zipcode.in_city.in_state.name ++ ' ' ++
                           .has_address.in_zipcode.digits
        }
        FILTER
            EXISTS (SELECT .in_category FILTER .name = <str>$category)
            AND
            .has_address.in_zipcode.digits = <str>$zip_code
        ORDER BY .review_count DESC
        LIMIT 3;
    """
    
    restaurants = client.query(business_query, category=category, zip_code=zip_code)
    
    print(f"Found {len(restaurants)} Italian restaurants:")
    for i, restaurant in enumerate(restaurants, 1):
        print(f"  {i}. {restaurant.name} ({restaurant.review_count} reviews)")
    print()
    
    # Now get reviews for each restaurant
    all_results = []
    
    for restaurant in restaurants:
        review_query = """
            SELECT Review {
                text,
                rating,
                written_by: {
                    user_id
                }
            }
            FILTER .reviews_business.ID = <str>$business_id
            LIMIT 3;
        """
        
        reviews = client.query(review_query, business_id=restaurant.ID)
        all_results.append({
            'restaurant': restaurant,
            'reviews': reviews
        })
    
    # Stop timing
    end_time = time.time()
    total_time = end_time - start_time
    
    # Show all the results
    print("Results:")
    print()
    
    for i, result in enumerate(all_results, 1):
        restaurant = result['restaurant']
        reviews = result['reviews']
        
        print(f"{i}. {restaurant.name}")
        print(f"   ID: {restaurant.ID}")
        print(f"   Address: {restaurant.full_address}")
        print(f"   Reviews: {restaurant.review_count} | Rating: {restaurant.rating}")
        print()
        
        print(f"   Review excerpts ({len(reviews)} found):")
        for j, review in enumerate(reviews, 1):
            # Get user name or use Anonymous
            user_name = "Anonymous"
            if review.written_by:
                user_name = f"User: {review.written_by.user_id}"
            
            # Make review text shorter
            review_text = review.text if review.text else "No review text"
            if len(review_text) > 120:
                review_text = review_text[:120] + "..."
            
            print(f"     Review {j} ({review.rating} stars) - {user_name}:")
            print(f"     \"{review_text}\"")
            print()
        
        print("-" * 50)
    
    print(f"Total time: {total_time:.3f} seconds")
    
    # Close connection
    client.close()


if __name__ == "__main__":
    find_italian_restaurants_with_reviews()
````
